## 1. Cài đặt thuật toán bằng NumPy

```python
import numpy as np
import matplotlib.pyplot as plt

def newton_raphson(grad_f, hess_f, x0, tol=1e-6, max_iter=50):
    x = np.array(x0, dtype=float)
    history = [x.copy()]
    
    for i in range(max_iter):
        g = grad_f(x)
        if np.linalg.norm(g) < tol:
            break
            
        H = hess_f(x)
        try:
            delta_x = np.linalg.solve(H, -g)
        except np.linalg.LinAlgError:
            print(f"Ma trận không khả nghịch tại bước {i}")
            break
            
        x = x + delta_x
        history.append(x.copy())
        
        if np.linalg.norm(delta_x) < tol:
            break
            
    return x, np.array(history)

def classify_extrema(H):
    eigvals = np.linalg.eigvals(H)
    if np.all(eigvals > 0):
        return "Cực tiểu"
    elif np.all(eigvals < 0):
        return "Cực đại"
    else:
        return "Saddle point"
```

## 2. Thử nghiệm và đánh giá

### 2.1. Thử nghiệm 1: Hàm bậc hai 


```python
def f1(x):
    return x[0]**2 + 2*x[1]**2 - 2*x[0]*x[1] - 4*x[0] + 6

def grad_f1(x):
    return np.array([2*x[0] - 2*x[1] - 4, 4*x[1] - 2*x[0]])

def hess_f1(x):
    return np.array([[2.0, -2.0], 
                     [-2.0, 4.0]])

x0_1 = [0.0, 0.0]
x_opt1, hist1 = newton_raphson(grad_f1, hess_f1, x0_1)

H1 = hess_f1(x_opt1)
loai1 = classify_extrema(H1)

print("--- KẾT QUẢ THỬ NGHIỆM 1 ---")
print(f"Điểm khởi tạo: {x0_1}")
print(f"Số bước lặp: {len(hist1) - 1}")
print(f"Điểm cực trị tìm được: x = {x_opt1[0]:.4f}, y = {x_opt1[1]:.4f}")
print(f"Giá trị f(x, y): {f1(x_opt1):.4f}")
print(f"Phân loại cực trị: {loai1}")
```

### 3.2. Thử nghiệm 2: Hàm phi tuyến nhiều điểm dừng

```python
def f2(x):
    return x[0]**4 + x[1]**4 - 4*x[0]*x[1] + 1

def grad_f2(x):
    return np.array([4*x[0]**3 - 4*x[1], 4*x[1]**3 - 4*x[0]])

def hess_f2(x):
    return np.array([[12.0*x[0]**2, -4.0], 
                     [-4.0, 12.0*x[1]**2]])

# Khởi tạo tại x0 = [2.0, 2.0]
x0_a = [2.0, 2.0]
x_opt_a, hist_a = newton_raphson(grad_f2, hess_f2, x0_a)
H_a = hess_f2(x_opt_a)

print("--- THỬ NGHIỆM 2 (Điểm xuất phát [2.0, 2.0]) ---")
print(f"Số bước lặp: {len(hist_a) - 1}")
print(f"Điểm cực trị: ({x_opt_a[0]:.4f}, {x_opt_a[1]:.4f})")
print(f"Giá trị hàm: {f2(x_opt_a):.4f}")
print(f"Phân loại: {classify_extrema(H_a)}\n")

# Khởi tạo tại x0 = [-2.0, -1.5]
x0_b = [-2.0, -1.5]
x_opt_b, hist_b = newton_raphson(grad_f2, hess_f2, x0_b)
H_b = hess_f2(x_opt_b)

print("--- THỬ NGHIỆM 2 (Điểm xuất phát [-2.0, -1.5]) ---")
print(f"Số bước lặp: {len(hist_b) - 1}")
print(f"Điểm cực trị: ({x_opt_b[0]:.4f}, {x_opt_b[1]:.4f})")
print(f"Giá trị hàm: {f2(x_opt_b):.4f}")
print(f"Phân loại: {classify_extrema(H_b)}")
```
