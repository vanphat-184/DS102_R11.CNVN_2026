# Lấy dữ liệu thông qua API

```python
import requests
import pandas as pd
import time

api_key = "QQbYGugwsVq5kcxFVVSe2APybHTO6j5dSuYyuSrp"
url = "https://api.fda.gov/drug/event.json"

data = []
for skip in range(0, 10000, 1000):
    params = {
        "api_key": api_key,
        "limit": 1000,
        "skip": skip,
        "search": "_exists_:(patient.patientonsetage AND patient.patientweight)"
    }
    res = requests.get(url, params=params).json().get("results", [])
    for item in res:
        p = item.get("patient", {})
        drugs = p.get("drug", [])
        reactions = p.get("reaction", [])
        data.append({
            "Age": p.get("patientonsetage"),
            "Weight_kg": p.get("patientweight"),
            "Sex": p.get("patientsex"),
            "Num_Drugs": len(drugs),
            "Indication": drugs[0].get("drugindication") if drugs else None,
            "Reaction": reactions[0].get("reactionmeddrapt") if reactions else None,
            "Hospitalization": item.get("seriousnesshospitalization", 0),
            "Death": item.get("seriousnessdeath", 0),
            "Disabling": item.get("seriousnessdisabling", 0)
        })
    time.sleep(0.5)

df = pd.DataFrame(data)
```

# Đọc dữ liệu và lưu thành file csv

```python
df.to_csv("openfda_drug_events.csv", index=False)
df.head(100)
```
