## 1. Giới thiệu bộ dữ liệu
- Bộ dữ liệu được thu thập từ **OpenFDA Drug Event API** (`https://api.fda.gov/drug/event.json`), ghi nhận thông tin về các biến cố bất lợi khi sử dụng thuốc.
- Dữ liệu khi thu thập có 9 columns và 10000 rows

**Ý nghĩa các trường thông tin:**
- `Age`: Tuổi của bệnh nhân.
- `Weight_kg`: Cân nặng của bệnh nhân (kg).
- `Sex`: Giới tính bệnh nhân (1: Nam, 2: Nữ, 0: Không xác định).
- `Num_Drugs`: Tổng số lượng thuốc bệnh nhân đang dùng.
- `Indication`: Chỉ định điều trị của thuốc.
- `Reaction`: Phản ứng phụ hoặc biến cố bất lợi gặp phải.
- `Hospitalization`: Biến cố gây nhập viện (1: Có, 0: Không).
- `Death`: Biến cố dẫn đến tử vong (1: Có, 0: Không).
- `Disabling`: Biến cố gây tàn tật, suy giảm chức năng (1: Có, 0: Không).
