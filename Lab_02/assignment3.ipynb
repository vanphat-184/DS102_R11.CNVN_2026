## 1. Chuẩn bị dữ liệu

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, precision_score, recall_score, f1_score

train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

features = ['Age', 'Weight_kg', 'Sex', 'Num_Drugs']

scaler = StandardScaler()
X_train = scaler.fit_transform(train[features])
X_test = scaler.transform(test[features])
```

## 2. Mô hình Logistic Regression (Phân loại nhị phân)

```python
y_train = train['Hospitalization'].values
y_test = test['Hospitalization'].values

clf_bin = LogisticRegression(max_iter=1000)
clf_bin.fit(X_train, y_train)

y_pred = clf_bin.predict(X_test)

print("--- Kết quả Logistic Regression ---")
print(f"Precision: {precision_score(y_test, y_pred, zero_division=0):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred, zero_division=0):.4f}")
print(f"F1-score:  {f1_score(y_test, y_pred, zero_division=0):.4f}")
```

## 3. Mô hình Softmax Regression (Phân loại đa lớp)

```python
y_train = train['Severity'].values
y_test = test['Severity'].values

clf_multi = LogisticRegression(max_iter=1000)
clf_multi.fit(X_train, y_train)

y_pred = clf_multi.predict(X_test)

print("--- Kết quả Softmax Regression ---")
print(f"Macro Precision: {precision_score(y_test, y_pred, average='macro', zero_division=0):.4f}")
print(f"Macro Recall:    {recall_score(y_test, y_pred, average='macro', zero_division=0):.4f}")
print(f"Macro F1-score:  {f1_score(y_test, y_pred, average='macro', zero_division=0):.4f}")
print("\nBáo cáo chi tiết:")
print(classification_report(y_test, y_pred, zero_division=0))
```
